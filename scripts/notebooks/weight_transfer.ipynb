```python
import torch 
```

```python
def process_model_state_dict(state_dict):
    model_dict = state_dict['model_state_dict']
    new_stae_dict = {}
    for k, v in model_dict.items():
        if k.startswith('actor'):
            if 'encoder' in k:
                continue
            elif ('actor_mlp' in k):
                new_stae_dict[k.replace('actor_mlp.','')] = v
        elif k.startswith('critic'):
            if 'encoder' in k:
                continue
            elif ('critic_mlp' in k):
                new_stae_dict[k.replace('critic_mlp.','')] = v
        elif k.startswith('encoder'):
            new_stae_dict[k.replace('encoder.','encoder.encoder.')] = v
        else:
            new_stae_dict[k] = v
    new_stae_dict['actor_obs_normalizer'] = state_dict['obs_norm_state_dict']
    return new_stae_dict
```

```python
state_dict = torch.load("model_5150.pt", map_location=torch.device('cpu'))
```

```python
new_state_dict = state_dict.copy()
new_state_dict["model_state_dict"] = process_model_state_dict(state_dict)
```

```python
torch.save(new_state_dict, "model_114514.pt")
```
